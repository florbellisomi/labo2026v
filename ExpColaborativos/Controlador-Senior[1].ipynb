```r
# Funciones del controlador. Se ejecutan en el mismo kernel R de esta notebook.
controlador <- local({
  tipos_base <- c("markdown", "markdown", "markdown", "markdown", "markdown", "markdown", "markdown", "code", "code", "code", "markdown", "code", "markdown", "code", "markdown", "markdown", "code", "markdown", "markdown", "code", "code", "code", "code", "code", "code", "code", "markdown", "code", "code", "code", "code", "code", "code", "code", "code", "code", "code", "code", "code", "code", "markdown", "code", "code", "code", "code", "code", "markdown", "markdown", "code", "code", "code", "code", "code", "code", "markdown", "code", "markdown", "code", "code", "code", "code", "code", "markdown", "markdown", "markdown", "code", "code", "code", "code", "code", "code", "markdown", "markdown", "markdown", "code", "code", "code", "code", "code", "markdown", "markdown", "markdown", "markdown", "code", "markdown", "code", "code", "markdown", "code", "markdown", "code", "markdown", "code", "markdown", "code", "markdown", "markdown", "markdown", "code", "markdown", "code", "code", "markdown", "code", "code", "code", "code", "code", "markdown", "markdown", "code", "code", "code", "code", "code", "code", "markdown", "markdown", "code", "code", "code")
  hashes_base <- c("34ef8b20bbc4a89686b254cad70749f4", "4d345c392d1b0ad861e6a7635b780d82", "badc3ff0b9289343758c47c152961c97", "3482cac10306f8600ba8d07919279806", "a04c855abfe0dc1f0f9ab58eef72d483", "48111bcb7aabac29ff8a6bd2dd5a441a", "93401b0388b478e9fff30bd053f07250", "a73cba134494ca4b63b8a40a518eecce", "e2584c34ad688f35ee9e910e47d41a24", "11a0c97d5c14d748cbf1ec7bec6205e3", "ee86ebca3f788b89b2c0617abc2d7910", "c30bd2baf19028292ea0fa04a65c641e", "4616e1852ba00f122b470ef71ac3966f", "761c707e7e640e2028b12772a31474f8", "17b8f5930ba9a7746380968de72359a4", "ba8072d047058a9ae276fb30605baf25", "683ce3494d3362b0a33f6dad703d68f9", "a8b7faecb74dcb1b075947579d526269", "0aa0c6deb045befc0fb1d1bd93a5bc85", "7787f95f3f27e8d2cad48b04c824d2e6", "d3f31ee6bec0344d752fbce2705f6be9", "2a3cf6c0bf49d66c2de6507c8461f41e", "09f97198aa04a756fd224a14220d2b0d", "60b89024a3c411b894bb0297a26d0440", "29d1cada11046acc7ffcf36cee07c43d", "58db6651adde3834fe7a597cf8391f8a", "5aaf9585efc6e8263e21c84f2cda8c56", "9ac29507f960551718b01441dc8f1ea2", "7bd2dcbd1fe326cae81424935257a86f", "86ee92a25e855da11df4a7d2c69366a3", "94420a6ec416f8c527736c8f97739534", "eb4d355a222156b06d6fcecf36c8b4c1", "48fbf185ef9928983fe433cfebe22d9f", "71d6beffaa2f7587b0c1ec27f8441972", "582338fa5cb209b37bd97190cfa081ae", "7a4ba8a1fa3f4bb1694058970a049c31", "00c76a04b48faa2a4a4cec93eb174a24", "890393b68ee3f07db8c4379c39b8de5a", "dcc7ab99ead35d688d55c0f5809813af", "3d190208059d4700002156e7aab219b2", "6a09135d40dc35afe2a8701ebffd0fb2", "73ea37d00f6575c5375239d34dfc5ae9", "890393b68ee3f07db8c4379c39b8de5a", "1388424504ec51193032a7534c959084", "6001c51ca7045cc838dd0897fbb479d0", "57b507d1a05a46c8710303716b9fa4c7", "6b122021693a1e60bd625433c59eb1b1", "e32ab830998711dbd6a133aef2b17b47", "143474813759b290c0130b15cb57be7d", "d7c711c8b1faf336ce03dec4a432277b", "3ef284cb415c014e7821aa907fd9d350", "a32d8bee726c085ee250c0081f691f3a", "1675f9e2a113a39320b4ff4b0dd0b5e8", "22acbd38111154d81bd324b26ab47271", "6ceb60d0aaaa9b6529c7ed55b8ae5be3", "de7c825940b9fce430c98dc20e0a5a3a", "a4a7617f11a20051e4fe8c7b938714d9", "29309e46920d96249d79ad86af08aa94", "0a6425f9001ee684e9dedb992f54d6fd", "bdee3651e4630d1615b33f551dce5727", "e3db0759139f2e92201ca7761e6a54f7", "8610919bee5972d844051835d34e66a9", "230fee17851423b63dfdf71fe5508c98", "98777e0cb6873d37c2fca6438b91d5b1", "7fbab302506aaaf769359c31c0b9ce66", "791c32b0304f7c28f6f89672928a1129", "c0967a853b10ec145c7d1a20e0141531", "7e49ef9daced1e0ea0b4b4c24259efc7", "a4d1ab4a510f9a10179043fad3837d99", "5667afe96062b5f274b83974b625a986", "246af779b6030698a3dee75bae0043c5", "a2129e1ee232da3b86a73d4be97b9495", "e1aa6404e8bd5ded994698aac98511de", "0cfabe57bbe8cc7af87f58ab5beb2fb4", "c313f3470d5e109b7dfb6815efe36ffb", "a60f892d937cc55508bdcdcdbf9baf11", "cfa0cab99b8bc03f1f47dc0c85cb466b", "34ef8b20bbc4a89686b254cad70749f4")
  evaluacion_b <- "# Adaptación B: controles previos a evaluación local.\nstopifnot(nrow(dfuture) >= 400, nrow(tb_prediccion) == nrow(dfuture),\n          !anyNA(dfuture$clase_ternaria),\n          all(dfuture$clase_ternaria %in% c(\"CONTINUA\", \"BAJA+1\", \"BAJA+2\")),\n          all(is.finite(tb_prediccion$prob)))\n# asigno clase ternaria\ntb_prediccion[, clase_ternaria := dfuture$clase_ternaria ]\n\n# asigno ganancias\ntb_prediccion[, ganancia := -0.025 ]\ntb_prediccion[clase_ternaria==\"BAJA+2\", ganancia := 0.975 ]\n\n# reordeno, y acumulada\nsetorder( tb_prediccion, -prob )\ntb_prediccion[, gan_acum := cumsum(ganancia)]\n\n# media movil de ancho 400\ntb_prediccion[,\n  gan_suavizada := frollmean(\n    x= gan_acum,\n    n= 400,\n    align= \"center\",\n    na.rm= TRUE,\n    hasNA= TRUE\n  )\n]\n\n# Este es la ganancia a reportar\nresultado <- list()\nresultado$ganancia_suavizada_max <- max( tb_prediccion$gan_suavizada, na.rm=TRUE )\noptions(digits= 8)\nresultado$envios <- which.max( tb_prediccion$gan_suavizada)\nresultado\nstopifnot(is.finite(resultado$ganancia_suavizada_max))\n# Agregados del controlador: persistencia y gráfico con escala de datos.\nfwrite(tb_prediccion, \"evaluacion_local.tsv\", sep=\"\\t\")\nfwrite(as.data.table(c(list(experimento=PARAM$experimento,\n  semilla=PARAM$semilla_primigenia, future=202107), resultado)),\n  \"resultado_local.tsv\", sep=\"\\t\")\nPARAM$out$evaluacion_local <- resultado\ntb_prediccion[, envios := .I]\npdf(\"curva_de_ganancia.pdf\")\nplot(tb_prediccion$envios, tb_prediccion$gan_acum, type=\"l\", col=\"gray\",\n     xlab=\"Clientes seleccionados\", ylab=\"Ganancia (unidades docentes)\",\n     main=paste(\"Grupo B - semilla\", PARAM$semilla_primigenia))\nlines(tb_prediccion$envios, tb_prediccion$gan_suavizada, col=\"blue\")\ndev.off()\ncapture.output(sessionInfo(), file=\"sessionInfo.txt\")\n"

  huella_texto <- function(texto) {
    archivo <- tempfile()
    on.exit(unlink(archivo))
    writeBin(charToRaw(enc2utf8(texto)), archivo)
    unname(tools::md5sum(archivo))
  }
  literal <- function(x) encodeString(x, quote='"')
  sustituir <- function(s, antes, despues) {
    posiciones <- gregexpr(antes, s, fixed=TRUE)[[1]]
    if (identical(posiciones[1], -1L) || length(posiciones) != 1L)
      stop("No coincide el bloque esperado: ", antes)
    sub(antes, despues, s, fixed=TRUE)
  }
  leer_base <- function(ruta) {
    if (!requireNamespace("jsonlite", quietly=TRUE))
      stop("El kernel R necesita jsonlite para leer notebooks. No se instaló nada automáticamente.")
    nb <- jsonlite::fromJSON(ruta, simplifyVector=FALSE)
    tipos <- vapply(nb$cells, function(c) c$cell_type, "")
    if (!identical(tipos, tipos_base)) stop("La estructura de celdas no coincide con la base revisada.")
    textos <- vapply(nb$cells, function(c) paste0(unlist(c$source), collapse=""), "")
    indices <- which(tipos == "code")
    distintos <- indices[vapply(textos[indices], huella_texto, "") != hashes_base]
    if (length(distintos)) stop("Código base distinto en celdas: ", paste(distintos, collapse=", "))
    textos[indices]
  }
  adaptar <- function(textos, cfg, semilla, experimento, carpeta) {
    indices <- which(tipos_base == "code")
    names(textos) <- as.character(indices)
    textos['12'] <- sustituir(textos['12'], "PARAM$semilla_primigenia <- 102191",
                             paste0("PARAM$semilla_primigenia <- ", semilla))
    textos['12'] <- sustituir(textos['12'], "PARAM$experimento <- 9500",
                             paste0("PARAM$experimento <- ", literal(experimento)))
    textos['14'] <- paste0("# Carpeta exclusiva creada por el controlador.\nsetwd(", literal(carpeta), ")")
    textos['17'] <- sustituir(textos['17'], 'paste0("/content/datasets/", PARAM$dataset)', literal(cfg$dataset))
    if (cfg$grupo == "B") {
      textos['75'] <- sustituir(textos['75'], "validate <- c(202107)", "validate <- c(202105)")
      textos['75'] <- sustituir(textos['75'], "202101, 202102, 202103, 202104, 202105", "202101, 202102, 202103")
      textos['99'] <- sustituir(textos['99'], "202101, 202102, 202103, 202104, 202105, 202106,\n  202107", "202101, 202102, 202103, 202104, 202105")
      textos['111'] <- sustituir(textos['111'], "future <- c(202109)", "future <- c(202107)")
      textos['119'] <- evaluacion_b
    } else if (!cfg$subir_kaggle) {
      textos['119'] <- sustituir(textos['119'], "Sys.sleep(30)", "# Envíos desactivados desde CONFIG.")
      textos['119'] <- sustituir(textos['119'], "salida <- system(linea, intern=TRUE)", 'salida <- "CSV guardado; envío a Kaggle desactivado"')
    } else {
      textos['119'] <- sustituir(textos['119'], "salida <- system(linea, intern=TRUE)", paste0(
        'salida <- system(linea, intern=TRUE)\n',
        '  if (!is.null(attr(salida, "status")) && attr(salida, "status") != 0)\n',
        '    stop("Falló el envío a Kaggle. Revisar console.log antes de repetir para evitar duplicados.")'))
    }
    textos
  }
  guardar_estado <- function(estado, carpeta) saveRDS(estado, file.path(carpeta, "estado_controlador.rds"))
  requeridos <- function(grupo) c("PARAM.yml", "prediccion.txt", "BO_log.txt", "sessionInfo.txt",
    if(grupo == "B") c("resultado_local.tsv", "evaluacion_local.tsv", "curva_de_ganancia.pdf"))

  preparar <- function(cfg, mostrar=TRUE) {
    if (!is.character(cfg$grupo) || length(cfg$grupo) != 1 || !cfg$grupo %in% c("A", "B")) stop("grupo debe ser A o B.")
    if (!is.logical(cfg$subir_kaggle) || length(cfg$subir_kaggle) != 1 || is.na(cfg$subir_kaggle)) stop("subir_kaggle debe ser TRUE o FALSE.")
    if (cfg$grupo == "B" && cfg$subir_kaggle) stop("Grupo B no permite envíos a Kaggle. Usar subir_kaggle=FALSE.")
    if (!is.numeric(cfg$semillas) || !length(cfg$semillas) || anyNA(cfg$semillas) ||
        any(!is.finite(cfg$semillas) | cfg$semillas <= 0 | cfg$semillas >= .Machine$integer.max | cfg$semillas != floor(cfg$semillas)) ||
        anyDuplicated(cfg$semillas)) stop("semillas debe contener enteros positivos, sin repetidos.")
    if (length(cfg$lote) != 1 || !grepl("^[A-Za-z0-9_-]+$", cfg$lote)) stop("lote admite letras, números, guion y guion bajo.")
    cfg$notebook_base <- normalizePath(path.expand(cfg$notebook_base), winslash="/", mustWork=TRUE)
    cfg$dataset <- normalizePath(path.expand(cfg$dataset), winslash="/", mustWork=TRUE)
    cfg$resultados <- normalizePath(path.expand(cfg$resultados), winslash="/", mustWork=TRUE)
    if (!dir.exists(cfg$resultados) || dir.exists(cfg$dataset)) stop("Revisar rutas de dataset y carpeta de resultados.")
    textos <- leer_base(cfg$notebook_base)
    names(textos) <- as.character(which(tipos_base == "code"))
    info <- file.info(cfg$dataset)
    plan <- lapply(cfg$semillas, function(semilla) {
      experimento <- paste0(cfg$grupo, "_", cfg$lote, "_s", format(semilla, scientific=FALSE, trim=TRUE))
      carpeta <- paste0(cfg$resultados, "/WF", experimento)
      bloques <- adaptar(textos, cfg, semilla, experimento, carpeta)
      script <- paste(paste0("# CELDA ORIGINAL ", names(bloques), "\n", bloques), collapse="\n\n")
      identidad <- list(version=2L, grupo=cfg$grupo, subir_kaggle=cfg$subir_kaggle,
        semilla=semilla, experimento=experimento, script_md5=huella_texto(script),
        dataset=cfg$dataset, dataset_bytes=info$size, dataset_mtime=as.numeric(info$mtime))
      estado <- "pendiente"
      if (file.exists(carpeta)) {
        ruta_estado <- file.path(carpeta, "estado_controlador.rds")
        if (!file.exists(ruta_estado)) stop("Carpeta ya existente sin estado: ", carpeta, ". Usar otro lote.")
        previo <- readRDS(ruta_estado)
        if (!identical(previo$identidad, identidad) || !identical(previo$estado, "completa"))
          stop("Corrida parcial o configuración distinta: ", carpeta, ". Revisar console.log; usar otro lote para repetir.")
        if (!all(file.exists(file.path(carpeta, requeridos(cfg$grupo))))) stop("Faltan salidas en ", carpeta)
        estado <- "completa: se omite"
      }
      if (mostrar) cat(experimento, "->", estado, "\n ", carpeta, "\n")
      list(carpeta=carpeta, identidad=identidad, estado=estado, bloques=bloques, script=script,
           celdas_modificadas=as.integer(names(textos)[textos != bloques]))
    })
    list(config=cfg, corridas=plan)
  }
  comprobar_paquetes <- function() {
    paquetes <- c("data.table", "R.utils", "mice", "lubridate", "Rcpp", "lightgbm", "DiceKriging", "mlrMBO", "primes", "yaml", "jsonlite")
    faltan <- paquetes[!vapply(paquetes, requireNamespace, logical(1), quietly=TRUE)]
    if (length(faltan)) stop("Este kernel R no carga: ", paste(faltan, collapse=", "),
      ". Bibliotecas: ", paste(.libPaths(), collapse="; "), ". Elegir el mismo kernel R donde funciona el baseline.")
  }
  correr_una <- function(corrida, grupo) {
    carpeta <- corrida$carpeta
    if (!dir.create(carpeta, showWarnings=FALSE)) stop("No se pudo reservar carpeta exclusiva: ", carpeta)
    writeLines(enc2utf8(corrida$script), file.path(carpeta, "workflow_ejecutado.R"), useBytes=TRUE)
    estado <- list(identidad=corrida$identidad, estado="en_curso", inicio=Sys.time(),
                   celdas_modificadas=corrida$celdas_modificadas)
    guardar_estado(estado, carpeta)
    wd_anterior <- getwd()
    opciones_anteriores <- options()
    # El baseline trabaja en este entorno, nunca en el del loop.
    entorno <- new.env(parent=globalenv())
    salida_antes <- sink.number()
    mensajes_antes <- sink.number(type="message")
    log <- file(file.path(carpeta, "console.log"), open="wt")
    terminado <- FALSE
    on.exit({
      if (!terminado) {
        estado$estado <- "interrumpida_o_fallida"
        estado$fin <- Sys.time()
        try(guardar_estado(estado, carpeta), silent=TRUE)
      }
      sink(type="message")
      if (mensajes_antes != 2L) sink(mensajes_antes, type="message")
      while (sink.number() > salida_antes) sink()
      close(log)
      setwd(wd_anterior)
      options(opciones_anteriores)
      rm(list=ls(envir=entorno, all.names=TRUE), envir=entorno)
      gc(verbose=FALSE)
    }, add=TRUE)
    sink(log, split=TRUE)
    sink(log, type="message")
    tryCatch({
      for (n in names(corrida$bloques)) {
        cat("\n--- Celda original ", n, " ---\n", sep="")
        for (expresion in parse(text=corrida$bloques[[n]])) {
          valor <- withVisible(eval(expresion, envir=entorno))
          if (valor$visible) print(valor$value)
        }
        flush(log)
      }
      capture.output(sessionInfo(), file=file.path(carpeta, "sessionInfo.txt"))
      if (!all(file.exists(file.path(carpeta, requeridos(grupo))))) stop("Faltan salidas esperadas.")
      if (grupo == "B") {
        resultado <- read.delim(file.path(carpeta, "resultado_local.tsv"))
        if (nrow(resultado) != 1 || !is.finite(resultado$ganancia_suavizada_max)) stop("Ganancia local inválida.")
        estado$resultado <- resultado
      }
      estado$estado <- "completa"
      estado$fin <- Sys.time()
      guardar_estado(estado, carpeta)
      terminado <- TRUE
    }, error=function(e) {
      estado$error <<- conditionMessage(e)
      stop(e)
    }, interrupt=function(e) {
      estado$error <<- "Interrupción del usuario"
      stop(e)
    })
    invisible(NULL)
  }
  ejecutar <- function(cfg) {
    plan <- preparar(cfg)
    comprobar_paquetes()
    if (cfg$grupo == "A" && cfg$subir_kaggle && !nzchar(Sys.which("kaggle"))) stop("No se encuentra el comando kaggle en este kernel.")
    for (corrida in plan$corridas) {
      if (corrida$estado != "pendiente") next
      cat("\nINICIANDO ", corrida$identidad$experimento, "\n", sep="")
      correr_una(corrida, cfg$grupo)
      cat("TERMINADA ", corrida$identidad$experimento, "\n", sep="")
    }
    cat("\nLista terminada. Las corridas completas previas se conservaron.\n")
    invisible(NULL)
  }
  list(preparar=preparar, ejecutar=ejecutar)
})
```
